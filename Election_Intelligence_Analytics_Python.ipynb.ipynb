## 5.1 Party-wise Total Votes

```python
party_votes = ( election_results
    .groupby("party")["votes"]
    .sum()
    .sort_values(ascending=False))

party_votes.plot(kind="bar", figsize=(25,15))
plt.title("Party-wise Total Votes",fontsize=40,color="red",fontweight="bold",y=1.15)
plt.xlabel("Party",fontsize=40,color="blue",fontweight="bold")
plt.ylabel("Total Votes",fontsize=40,color="black",fontweight="bold",y=0.5)
plt.xticks(rotation=90,fontsize=10)
plt.yticks(fontsize=10)
plt.tight_layout()
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.show()
```

### Observation:
### The chart compares the total votes secured by each political party, highlighting the parties with the strongest overall vote share.

## 5.2 Party-wise Seat Count

```python
winners = (
    election_results
    .sort_values("votes", ascending=False)
    .groupby(["election_year", "ac_number"])
    .first()
    .reset_index()
)

seat_count = (
    winners
    .groupby("party")
    .size()
    .sort_values(ascending=False)
)

seat_count.plot(kind="bar", figsize=(40,15))

plt.title("Party-wise Seat Count", fontsize=40, color="red", fontweight="bold", y=1.05)
plt.xlabel("Political Party", fontsize=35, color="blue", fontweight="bold")
plt.ylabel("Seats Won", fontsize=35, color="black", fontweight="bold")
plt.xticks(rotation=90, fontsize=10)
plt.yticks(fontsize=10)
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

### Observation:
### Displays the number of constituencies won by each political party.

## 5.3 Top 10 Candidates by Votes

```python
top10 = (
    election_results
    .sort_values("votes", ascending=False)
    .head(10)
)

top10.plot(
    x="candidate",
    y="votes",
    kind="bar",
    figsize=(40,15)
)

plt.title("Top 10 Candidates by Votes", fontsize=40, color="red", fontweight="bold", y=1.05)
plt.xlabel("Candidate", fontsize=35, color="blue", fontweight="bold")
plt.ylabel("Votes", fontsize=35, color="black", fontweight="bold")
plt.xticks(rotation=90, ha="right", fontsize=10)
plt.yticks(fontsize=10)
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

### Observation:
### Shows the candidates who secured the highest number of votes.

## 5.5 Average Turnout by Election Year

```python
election_results.info()
```

```python
print(election_results["turnout"].dtype)
```

```python
election_results["turnout"] = pd.to_numeric(
    election_results["turnout"],
    errors="coerce"
)
```

```python
print(election_results["turnout"].dtype)
```

```python
turnout = (
    election_results
    .groupby("election_year")["turnout"]
    .mean()
)

turnout.plot(kind="bar", figsize=(40,10))

plt.title("Average Voter Turnout by Election Year", fontsize=35, color="red", fontweight="bold")
plt.xlabel("Election Year", fontsize=25, color="blue", fontweight="bold")
plt.ylabel("Average Turnout (%)", fontsize=25, color="black", fontweight="bold")
plt.xticks(rotation=0, fontsize=20)
plt.yticks(fontsize=20)
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

### Observation:
### Compares the average voter turnout across election years.

## 5.6 Average Turnout by Region

```python
region_turnout = (
    constituency
    .merge(election_results, on="ac_number")
    .groupby("region_x")["turnout"]
    .mean()
)

region_turnout.plot(kind="bar", figsize=(40,15))

plt.title("Average Turnout by Region", fontsize=40, color="red", fontweight="bold")
plt.xlabel("Region", fontsize=35, color="blue", fontweight="bold")
plt.ylabel("Average Turnout (%)", fontsize=35, color="black", fontweight="bold")
plt.xticks(rotation=45, fontsize=20)
plt.yticks(fontsize=20)
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

### Observation:
### Highlights regional differences in average voter turnout.

## 5.7 Largest Victory Margins

```python
winner = (
    election_results
    .sort_values(["election_year","ac_number","votes"], ascending=[True,True,False])
)

winner["rank"] = winner.groupby(["election_year","ac_number"]).cumcount()+1

winner1 = winner[winner["rank"]==1]
winner2 = winner[winner["rank"]==2]

margin = winner1.copy()
margin["victory_margin"] = winner1["votes"].values - winner2["votes"].values

largest = (
    margin
    .sort_values("victory_margin", ascending=False)
    .head(10)
)

largest.plot(
    x="candidate",
    y="victory_margin",
    kind="bar",
    figsize=(40,15)
)

plt.title("Largest Victory Margins", fontsize=40, color="red", fontweight="bold")
plt.xlabel("Candidate", fontsize=35, color="blue", fontweight="bold")
plt.ylabel("Victory Margin", fontsize=35, color="black", fontweight="bold")
plt.xticks(rotation=45, ha="right", fontsize=18)
plt.yticks(fontsize=20)
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

### Observation:
### Displays the constituencies with the highest winning margins.

## 5.8 Closest Electoral Contests

```python
closest = (
    margin
    .sort_values("victory_margin")
    .head(10)
)

closest.plot(
    x="candidate",
    y="victory_margin",
    kind="bar",
    figsize=(40,15)
)

plt.title("Closest Electoral Contests", fontsize=40, color="red", fontweight="bold")
plt.xlabel("Candidate", fontsize=35, color="blue", fontweight="bold")
plt.ylabel("Victory Margin", fontsize=35, color="black", fontweight="bold")
plt.xticks(rotation=45, ha="right", fontsize=18)
plt.yticks(fontsize=20)
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

### Observation:
### Highlights the constituencies where the difference between the winner and runner-up was the smallest.
